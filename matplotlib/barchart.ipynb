```python
import matplotlib.pyplot as plt

movies = [
    "the dark knight",
    "love",
    "means",
    "you",
    "and"
]
year = [2000,2005,2007,2012,2025]
revenue = [245,190,100,120,700]

plt.bar(year,revenue,color = "black")
plt.title("Movie revenue")
plt.xlabel("year")
plt.ylabel("revenue")
for i in range(len(year)):
    plt.text(year[i],revenue[i]+10,str(revenue[i]), ha = "center")

plt.ylim(0,max(revenue)+200)

plt.show()
```

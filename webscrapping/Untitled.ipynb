```python
import requests 

URL = "https://quotes.toscrape.com/"
res = requests.get(URL)
```

```python
print(res.text)
```

```python
with open("scraped_data/web.html","w") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup 

with open("scraped_data/web.html","r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content,"lxml")
```

```python
soup.find("h3")
soup.find_all("h1")
```

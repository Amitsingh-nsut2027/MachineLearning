```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.get_dataset_names()
flight_data = sns.load_dataset('flights')
# print(flight_data)

flight = flight_data.pivot(index = "month" , columns = "year", values = "passengers")
print(flight)

sns.heatmap(
    flight,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)
plt.title("year and passengers data")
plt.tight_layout()
      

# sns.heatmap()
```
